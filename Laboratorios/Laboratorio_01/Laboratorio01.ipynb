```python
archivo = "/content/drive/MyDrive/Colab Notebooks/Machine Learning Soto/Datasets/Lab01/auto_mpg_peso_mpg.csv"

x = []
y = []

with open(archivo, "r") as dataset:

    # Saltamos la primera línea porque contiene:
    # weight,mpg
    next(dataset)

    for linea in dataset:

        datos = linea.strip().split(",")

        peso = float(datos[0])
        mpg = float(datos[1])

        x.append(peso)
        y.append(mpg)

print("Cantidad de datos:", len(x))
```

```python
b = 40
w = -0.006

peso = x[0]

prediccion = b + w * peso

print("Peso:", peso)
print("MPG real:", y[0])
print("MPG predicho:", prediccion)
```

```python
error = abs(y[0] - prediccion)

print("Error absoluto:", error)
```

Aqui hago una busqueda exaustiva que encuentre el mejor error para poder usarlo para inferir nuevos resultados, es decir mejor "y" usando una condicional de acuerdo a un error promedio obtenido.

```python
mejor_error = 999999999

mejor_b = 0
mejor_w = 0


for b in range(0, 61):

    for numero_w in range(-200, 1):

        w = numero_w / 10000

        error_total = 0


        for i in range(len(x)):

            prediccion = b + w * x[i]

            error = abs(y[i] - prediccion)

            error_total = error_total + error


        error_promedio = error_total / len(x)


        if error_promedio < mejor_error:

            mejor_error = error_promedio
            mejor_b = b
            mejor_w = w


print("Mejor valor de b:", mejor_b)
print("Mejor valor de w:", mejor_w)
print("Mejor error promedio:", mejor_error)
```

```python
mejor_error = 999999999

mejor_b = 0
mejor_w = 0


# Probamos b desde 40.0 hasta 48.0
for numero_b in range(400, 481):

    b = numero_b / 10


    # Probamos w desde -0.00900 hasta -0.00500
    for numero_w in range(-900, -499):

        w = numero_w / 100000

        error_total = 0


        for i in range(len(x)):

            prediccion = b + w * x[i]

            error = abs(y[i] - prediccion)

            error_total = error_total + error


        error_promedio = error_total / len(x)


        if error_promedio < mejor_error:

            mejor_error = error_promedio
            mejor_b = b
            mejor_w = w


print("Mejor b:", mejor_b)
print("Mejor w:", mejor_w)
print("Error promedio:", mejor_error)
```

```python
nuevos_x = [
    1750,
    2250,
    2750,
    3000,
    3500,
    3750,
    4000,
    4250,
    4500,
    4750
]

print("INFERENCIAS")
print("---------------------------")

for nuevo_x in nuevos_x:

    prediccion = mejor_b + mejor_w * nuevo_x

    print(
        "Peso:",
        nuevo_x,
        "libras -> MPG estimado:",
        round(prediccion, 2)
    )
```
